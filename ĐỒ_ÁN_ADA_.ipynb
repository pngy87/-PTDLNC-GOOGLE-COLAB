**1. XÁC ĐỊNH DỮ LIỆU THIẾU**

```python
# Tạo một danh sách các cặp (Tên hiển thị, DataFrame)
datasets = [
    ("Flights", df_flights),
    ("Airlines", df_airlines),
    ("Airports", df_airports)
]

for name, df in datasets:
    null_counts = df.isnull().sum()
    null_columns = null_counts[null_counts > 0].sort_values(ascending=False)

    print(f"\n[{name}]")
    if len(null_columns) > 0:
        print(null_columns)
    else:
        print("Không có giá trị null.")
```

**2. XÁC ĐỊNH DỮ LIỆU TRÙNG LẶP**

```python
for name, df in datasets:
    print(f"KIỂM TRA BỘ DỮ LIỆU: {name.upper()}")
    print(f"{'='*30}")

    # Kiểm tra từng dòng có bị trùng lặp không?
    dup_rows_count = df.duplicated().sum()
    print(f"1. Số lượng dòng trùng lặp hoàn toàn: {dup_rows_count}")

    if dup_rows_count > 0:
        print(f" Các dòng trùng lặp (5 dòng đầu):")
        display(df[df.duplicated(keep=False)].head(5))
    else:
        print(" Không có dòng nào bị trùng lặp hoàn toàn.")

    # Kiểm tra từng cột có bị trùng lặp không?
    unique_cols_data = {}
    dup_data_cols = []

    for col in df.columns:
        col_data = tuple(df[col].iloc[:10000].values)

        if col_data in unique_cols_data:
            dup_data_cols.append((unique_cols_data[col_data], col))
        else:
            unique_cols_data[col_data] = col

    print(f"\n 2. Kiểm tra trùng lặp dữ liệu giữa các cột:")
    if dup_data_cols:
        print(f" Tìm thấy {len(dup_data_cols)} cặp cột giống hệt nhau:")
        display(pd.DataFrame(dup_data_cols, columns=['Cột Gốc', 'Cột Trùng']))
    else:
        print("Không tìm thấy cột nào có dữ liệu giống hệt nhau.")
```

# **TIỀN XỬ LÝ DỮ LIỆU**

**1. Làm sạch dữ liệu**

```python
flights = pd.read_csv('flights.csv', low_memory=False)
airlines = pd.read_csv('airlines.csv')
airports = pd.read_csv('airports.csv')

# Lọc bỏ chuyến hủy/đổi hướng
flights_clean = flights[(flights['CANCELLED'] == 0) & (flights['DIVERTED'] == 0)].copy()

# Xóa cột thừa để nhẹ bộ nhớ
cols_to_drop = ['CANCELLED', 'CANCELLATION_REASON', 'DIVERTED']
flights_clean.drop(columns=cols_to_drop, inplace=True, errors='ignore')

# Điền 0 cho các nguyên nhân trễ (Business Logic cho Diagnostic)
delay_cols = ['AIR_SYSTEM_DELAY', 'SECURITY_DELAY', 'AIRLINE_DELAY',
              'LATE_AIRCRAFT_DELAY', 'WEATHER_DELAY']
flights_clean[delay_cols] = flights_clean[delay_cols].fillna(0)

# Loại bỏ dòng thiếu giờ thực tế
flights_clean.dropna(subset=['ARRIVAL_TIME', 'DEPARTURE_TIME'], inplace=True)

print(f"Kích thước sau khi lọc: {flights_clean.shape}")
```

**2. Gộp và tạo feature**

```python
# Chuyển đổi thời gian
flights_clean['DATE'] = pd.to_datetime(flights_clean[['YEAR', 'MONTH', 'DAY']])

# Gộp tên Hãng bay & Sân bay
flights_full = pd.merge(flights_clean, airlines, left_on='AIRLINE', right_on='IATA_CODE', how='left')
flights_full.rename(columns={'AIRLINE_y': 'AIRLINE_NAME'}, inplace=True)

flights_full = pd.merge(flights_full, airports[['IATA_CODE', 'AIRPORT', 'CITY', 'STATE']],
                        left_on='ORIGIN_AIRPORT', right_on='IATA_CODE', how='left')
flights_full.rename(columns={'AIRPORT': 'ORIGIN_AIRPORT_NAME'}, inplace=True)

# Xóa cột mã thừa
flights_full.drop(['IATA_CODE_x', 'IATA_CODE_y', 'IATA_CODE'], axis=1, errors='ignore', inplace=True)

# Phân loại mức độ trễ (Delay Level)
def classify_delay(minutes):
    if minutes <= 15: return 'On-time'
    elif minutes <= 60: return 'Short Delay'
    elif minutes <= 180: return 'Medium Delay'
    else: return 'Long Delay'

flights_full['DELAY_LEVEL'] = flights_full['ARRIVAL_DELAY'].apply(classify_delay)

# Biến Cuối tuần (Is Weekend)
flights_full['IS_WEEKEND'] = flights_full['DAY_OF_WEEK'].apply(lambda x: 1 if x >= 6 else 0)

# Khung giờ bay (Time Block)
def get_hour(time_num):
    if pd.isnull(time_num): return np.nan
    str_time = str(int(time_num)).zfill(4)
    return int(str_time[:2])

flights_full['DEPARTURE_HOUR'] = flights_full['SCHEDULED_DEPARTURE'].apply(get_hour)
flights_full['TIME_BLOCK'] = pd.cut(flights_full['DEPARTURE_HOUR'],
                                    bins=[-1, 4, 11, 17, 24],
                                    labels=['Night', 'Morning', 'Afternoon', 'Evening'])
```

## **4. Tổng hợp & Tóm tắt dữ liệu (Aggregation & Summarization)**

### *4.1. Tổng hợp theo tháng*

```python
monthly_summary = flights_full.groupby(
    flights_full['DATE'].dt.to_period('M')
).agg({
    'ARRIVAL_DELAY':'mean',
    'DEPARTURE_DELAY':'mean',
    'DISTANCE':'mean'
})

monthly_summary.head()
```

### *4.2. Pivot table – Delay theo hãng & khung giờ*

```python
pivot_delay = pd.pivot_table(
    flights_full,
    values='ARRIVAL_DELAY',
    index='AIRLINE_NAME',
    columns='TIME_BLOCK',
    aggfunc='mean'
)

pivot_delay
```
